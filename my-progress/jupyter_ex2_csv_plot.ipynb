# 2-mashq: CSV, DataFrame va grafik

Ma'lumot: **GPU Setup** darsida o'lchagan CPU va GPU natijalarim.

```python
import matplotlib.pyplot as plt
import pandas as pd

csv_path = "benchmark.csv"
```

## CSV'ni yuklash

```python
df = pd.read_csv(csv_path)
df
```

## Grafik: matritsa o'lchami bo'yicha vaqt

```python
%matplotlib inline

table = df.pivot(index="size", columns="device", values="seconds")
table.plot(kind="bar", figsize=(6, 3), title="GT 1030: CPU vs GPU", ylabel="soniya")
plt.show()
```

```python
table["speedup"] = (table["CPU"] / table["GPU"]).round(2)
table
```
