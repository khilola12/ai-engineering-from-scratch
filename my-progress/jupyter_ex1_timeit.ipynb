# 1-mashq: list comprehension vs NumPy

100 000 ta tasodifiy son yaratishning ikki usulini `%timeit` bilan solishtiramiz.

```python
import random

import numpy as np

N = 100_000
```

## Oddiy Python: list comprehension

```python
%timeit [random.random() for _ in range(N)]
```

## NumPy

```python
%timeit np.random.rand(N)
```

**Xulosa:** NumPy sonlarni C tilida, bitta blokda yaratadi. Python esa har bir sonni alohida obyekt qilib, sikl bilan yaratadi.
